```python
import csv, os
from pathlib import Path

DATA_ROOTS = [
    Path('/home/igor/igor_repos/scaling_laws/data_local'),
    Path('/home/igor/igor_repos/scaling_laws/data'),
]
csv_path = 'missing_mi_on_disk.csv'

missing = 0
found = 0

with open(csv_path) as f:
    for row in csv.DictReader(f):
        suffix = '_geneformer' if row['algorithm'] == 'Geneformer' else ''
        stem = f"Y_{row['signal']}_{row['quality']}{suffix}"
        rel = Path(row['dataset']) / row['size'] / row['quality'] / 'results' / row['algorithm'] / 'model' / 'MI' / row['seed'] / stem / 'lmi_mutual_information.txt'
        if any((root / rel).exists() for root in DATA_ROOTS):
            found += 1
        else:
            missing += 1

total = found + missing
print(f"Done:    {found} / {total}")
print(f"Missing: {missing} / {total}")
```
